# ¿Qué haremos?

Trabajaremos con los datos del equipo 01-10 para explorar relaciones entre variables numéricas.

El coeficiente de Pearson (`r`) mide la fuerza de una relación lineal:

- `r` cercano a `1`: relación positiva fuerte.
- `r` cercano a `-1`: relación negativa fuerte.
- `r` cercano a `0`: poca relación lineal.

````python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


data_path = 'https://raw.githubusercontent.com/UNAM-FESAc/trab2026-modulo5/main/data/activity_data_team_01.csv'
df = pd.read_csv(data_path)
df.head()
````

## 1. Calcular la matriz de Pearson

Calcula la matriz de correlación de las columnas numéricas usando el método `pearson`. Guarda el resultado en `correlation_matrix`.

Copia y pega el código y completa las líneas que hacen falta:


```python
numeric_data = df.select_dtypes(include='number')
correlation_matrix = __________________.corr(method='pearson')
correlation_matrix.round(2)
```

````python
# answer 1
numeric_data = df.select_dtypes(include='number')
correlation_matrix = numeric_data.corr(method='pearson')
correlation_matrix.round(2)
````

Crea un mapa de colores con `imshow`. Guarda el eje en `ax`, agrega una barra de color y muestra los nombres de las variables en ambos ejes.

````python
# answer 2
fig, ax = plt.subplots(figsize=(8, 6))
image = ax.imshow(correlation_matrix, cmap='coolwarm', vmin=-1, vmax=1)
fig.colorbar(image, ax=ax, label='Pearson r')
ax.set_xticks(range(len(correlation_matrix.columns)))
ax.set_yticks(range(len(correlation_matrix.columns)))
ax.set_xticklabels(correlation_matrix.columns, rotation=45, ha='right')
ax.set_yticklabels(correlation_matrix.columns)
ax.set_title('Matriz de correlación de Pearson')
plt.tight_layout()
plt.show()
````

## 3. Encontrar la relación más fuerte

Ignora la diagonal de la matriz y guarda en `strongest_pair` una tupla con los nombres de las dos variables que tienen la mayor correlación absoluta.

Copia, pega el código y completa la(s) línea(s) que hacen falta:

```python
correlations_without_diagonal = correlation_matrix.where(~np.eye(__________________.shape[0], dtype=bool))
strongest_pair = correlations_without_diagonal.abs().stack().idxmax()
strongest_pair
```

````python
# answer 3
correlations_without_diagonal = correlation_matrix.where(~np.eye(correlation_matrix.shape[0], dtype=bool))
strongest_pair = correlations_without_diagonal.abs().stack().idxmax()
strongest_pair
````
