````python
from pathlib import Path

import pandas as pd

data_url = 'https://raw.githubusercontent.com/UNAM-FESAc/trab2026-modulo5/main/data/activity_data.csv'
df = pd.read_csv(data_url)
df.head()
````

## 1. Revisar el tamaño de los datos

Guarda el número de filas en `numero_filas` y el número de columnas en `numero_columnas`.

````python
# answer 1
numero_filas = df.shape[0]
numero_columnas = df.shape[1]
print(numero_filas, numero_columnas)
````

## 2. Calcular una media

Calcula la media de `measurement_before` y guárdala en `media_inicial`.

````python
# Answer 2
media_inicial = df['measurement_before'].mean()
print(media_inicial)
````

## 3. Contar participantes por grupo

Cuenta cuántas observaciones pertenecen a cada valor de `group` y guarda el resultado en `conteo_grupos`.

````python
# answer 3
conteo_grupos = df['group'].value_counts()
print(conteo_grupos)
````
