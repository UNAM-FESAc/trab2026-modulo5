# Módulo 5: Ejemplos Prácticos 📊

En este cuaderno verás ejemplos prácticos de cómo usar Python para análisis de datos.

**Objetivo**: Aprender a trabajar con datos reales usando bibliotecas como Pandas.

## Ejemplo 1: Cargar y Explorar Datos

Lo primero que hacemos es cargar nuestros datos desde un archivo CSV.

```python
# Importar la biblioteca Pandas
import pandas as pd
import numpy as np

# Cargar los datos desde el archivo CSV
df = pd.read_csv('../data/investigacion_datos.csv')

# Mostrar las primeras filas
print("Primeras 5 filas de nuestros datos:")
print(df.head())
```

### Explicación:
- `import pandas as pd` - Importamos la biblioteca Pandas y le damos el alias "pd"
- `pd.read_csv()` - Lee un archivo CSV (valores separados por comas)
- `df.head()` - Muestra las primeras 5 filas (df = DataFrame, un tipo especial de tabla)

## Ejemplo 2: Información Básica del Dataset

```python
# ¿Cuántas filas y columnas tenemos?
filas, columnas = df.shape
print(f"El dataset tiene:")
print(f"  - {filas} filas (registros)")
print(f"  - {columnas} columnas")
```

```python
# ¿Cuáles son los nombres de las columnas?
print("\nNombres de las columnas:")
print(df.columns.tolist())
```

```python
# Información detallada sobre el dataset
print("\nInformación detallada:")
print(df.info())
```

## Ejemplo 3: Estadísticas Básicas

```python
# Estadísticas numéricas de nuestros datos
print("Estadísticas básicas:")
print(df.describe())
```

### Interpretación:
- **count** - Cantidad de valores
- **mean** - Promedio
- **std** - Desviación estándar (variabilidad)
- **min** - Valor mínimo
- **25%** - Primer cuartil
- **50%** - Mediana (valor del medio)
- **75%** - Tercer cuartil
- **max** - Valor máximo

## Ejemplo 4: Operaciones Simples en Columnas

```python
# Acceder a una columna específica
print("Columna 'valor':")
print(df['valor'])
```

```python
# Calcular el promedio
promedio = df['valor'].mean()
print(f"Valor promedio: {promedio}")
```

```python
# Otras operaciones útiles
print(f"Valor máximo: {df['valor'].max()}")
print(f"Valor mínimo: {df['valor'].min()}")
print(f"Suma total: {df['valor'].sum()}")
print(f"Desviación estándar: {df['valor'].std()}")
```

## Ejemplo 5: Filtrar Datos

Muchas veces queremos trabajar solo con una parte de los datos.

```python
# Filtrar por categoría
print("Datos de tipo A:")
tipo_a = df[df['categoria'] == 'TipoA']
print(tipo_a)
```

```python
# Filtrar por valor
print("\nValores mayores a 120:")
valores_altos = df[df['valor'] > 120]
print(valores_altos)
```

```python
# Combinar filtros
print("\nTipo A con valor mayor a 110:")
resultado = df[(df['categoria'] == 'TipoA') & (df['valor'] > 110)]
print(resultado)
```

## Ejemplo 6: Agrupar y Resumir Datos

```python
# Agrupar por categoría y calcular el promedio
print("Promedio de valor por categoría:")
promedio_por_categoria = df.groupby('categoria')['valor'].mean()
print(promedio_por_categoria)
```

```python
# Contar cuántos elementos hay por categoría
print("\nCuántos elementos hay por categoría:")
cantidad_por_categoria = df.groupby('categoria').size()
print(cantidad_por_categoria)
```

```python
# Resumen múltiple
print("\nResumen por categoría:")
resumen = df.groupby('categoria')['valor'].agg(['count', 'mean', 'min', 'max', 'sum'])
print(resumen)
```

## Ejemplo 7: Crear Nuevas Columnas

```python
# Crear una nueva columna con valores transformados
df['valor_duplicado'] = df['valor'] * 2
print("Nueva columna 'valor_duplicado':")
print(df[['nombre', 'valor', 'valor_duplicado']])
```

```python
# Crear una columna basada en condición
df['clasificacion'] = df['valor'].apply(lambda x: 'Alto' if x > 140 else 'Bajo')
print("\nNueva columna 'clasificacion':")
print(df[['nombre', 'valor', 'clasificacion']])
```

## Ejemplo 8: Ordenar Datos

```python
# Ordenar por valor (ascendente)
print("Datos ordenados por valor (menor a mayor):")
print(df.sort_values('valor'))
```

```python
# Ordenar por valor (descendente)
print("\nDatos ordenados por valor (mayor a menor):")
print(df.sort_values('valor', ascending=False))
```

## Ejemplo 9: Trabajar con Valores Faltantes

```python
# Detectar valores faltantes (NaN)
print("¿Hay valores faltantes?")
print(df.isnull().sum())
```

```python
# Rellenar valores faltantes según el tipo de cada columna
df_con_relleno = df.copy()

for columna in df_con_relleno.columns:
    if pd.api.types.is_numeric_dtype(df_con_relleno[columna]):
        # Las columnas numéricas se rellenan con su media.
        df_con_relleno[columna] = df_con_relleno[columna].fillna(
            df_con_relleno[columna].mean()
        )
    else:
        # Las columnas de texto o categóricas se rellenan con su moda.
        moda = df_con_relleno[columna].dropna().mode()
        if not moda.empty:
            df_con_relleno[columna] = df_con_relleno[columna].fillna(moda.iloc[0])

print("Datos después de rellenar valores faltantes:")
print(df_con_relleno)
print("\nValores faltantes restantes por columna:")
print(df_con_relleno.isnull().sum())
```

## Ejemplo 10: Visualización Básica

```python
import matplotlib.pyplot as plt

# Gráfico simple de barras
df_resumido = df.groupby('categoria')['valor'].mean()
df_resumido.plot(kind='bar', title='Valor Promedio por Categoría')
plt.xlabel('Categoría')
plt.ylabel('Valor Promedio')
plt.show()
```

## Ejemplo 11: Contar Valores Únicos

```python
# ¿Cuántas categorías diferentes hay?
print("Categorías únicas:")
print(df['categoria'].unique())
print(f"\nTotal de categorías: {df['categoria'].nunique()}")
```

```python
# Contar ocurrencias de cada categoría
print("\nCuántos datos hay por categoría:")
print(df['categoria'].value_counts())
```
