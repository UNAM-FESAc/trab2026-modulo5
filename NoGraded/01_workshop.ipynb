## 5. Crear una nueva variable

Nuestra pregunta no se refiere solamente a las mediciones iniciales y posteriores. Nos interesa el **cambio**.

Matemáticamente:

\[
\Delta_i = y_i^{(\mathrm{after})} - y_i^{(\mathrm{before})}.
\]

Ahora traduciremos esa definición a código.

````python
# Calculamos el cambio para cada observación.
data["change"] = (
    data["measurement_after"]
    - data["measurement_before"]
)

data.head()
````

Observa que apareció una nueva columna: `change`.

Acabamos de realizar una operación fundamental en investigación computacional:

**concepto científico → definición matemática → transformación de datos**

## 6. Pregunta de investigación 1

### ¿Cambió la medición después de la intervención?

Calculemos el cambio promedio.

````python
# Calculamos el promedio del cambio observado.
mean_change = data["change"].mean()

mean_change
````

### 🔎 Interpreta antes de continuar

1. ¿El cambio promedio es positivo, negativo o cercano a cero?
2. ¿Qué sugiere este resultado?
3. ¿Un promedio positivo demuestra que la intervención **causó** el cambio?

La tercera pregunta es especialmente importante: **un cálculo descriptivo no demuestra causalidad**.

## 7. Pregunta de investigación 2

### ¿El cambio es diferente entre los grupos A y B?

Agruparemos las observaciones por `group` y calcularemos el cambio promedio.

````python
# Comparamos el cambio promedio de los dos grupos.
mean_change_by_group = data.groupby("group")["change"].mean()

mean_change_by_group
````

### 🔎 Detente e interpreta

- ¿Qué grupo presenta el mayor cambio promedio?
- ¿La diferencia observada es suficiente para afirmar que un grupo es mejor?
- ¿Qué otra información necesitaríamos antes de hacer una afirmación científica fuerte?

El computador puede calcular la diferencia. **La persona investigadora debe decidir qué puede concluirse de ella.**

## 8. Visualizar la variabilidad

Los promedios pueden ocultar información importante. Veamos cómo se distribuye `change`.

````python
import matplotlib.pyplot as plt

# Visualizamos la distribución del cambio observado.
plt.hist(data["change"])

plt.xlabel("Observed change")
plt.ylabel("Frequency")
plt.title("Distribution of observed changes")

plt.show()
````

### 🔎 Observa la figura

¿Todos los participantes muestran el mismo cambio?

Esta figura introduce una idea fundamental: **la variabilidad también forma parte de la evidencia**.

## 9. Pregunta de investigación 3

### ¿Existe una asociación entre el tiempo de exposición y el cambio observado?

Una gráfica de dispersión permite examinar visualmente la relación entre dos variables.

````python
# Visualizamos la relación entre tiempo de exposición y cambio.
plt.scatter(
    data["hours"],
    data["change"]
)

plt.xlabel("Exposure time (hours)")
plt.ylabel("Observed change")
plt.title("Exposure time vs. observed change")

plt.show()
````

### 🔎 Antes de calcular nada

Observa la nube de puntos.

- ¿Parece existir una tendencia?
- ¿La relación parece positiva o negativa?
- ¿Hay variabilidad?
- ¿Puedes identificar observaciones inusuales?

Primero **observamos**. Después calculamos.

## 10. Medir la asociación

Podemos calcular el coeficiente de correlación lineal entre `hours` y `change`.

\[
-1 \leq r \leq 1
\]

De manera simplificada:

- \(r>0\): asociación lineal positiva.
- \(r<0\): asociación lineal negativa.
- \(rpprox0\): poca asociación lineal.

La magnitud debe interpretarse en el contexto del problema.

````python
# Calculamos la correlación lineal.
correlation = data["hours"].corr(data["change"])

correlation
````

## 🧠 Pausa científica

El computador acaba de calcular un número.
