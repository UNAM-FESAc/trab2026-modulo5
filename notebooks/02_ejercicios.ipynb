## Ejercicio 3.1: Análisis Completo

**Tarea:** Escribe un programa que:
1. Calcule estadísticas por categoría (promedio, máximo, mínimo)
2. Identifique cuál es el valor máximo global
3. Identifique cuál es la muestra (nombre) con el valor máximo
4. Imprime un resumen en formato legible

````python
# Escribe tu código aquí
````

**Solución esperada:**

````python
# Solución
print("=== ANÁLISIS POR CATEGORÍA ===")
for categoria in df['categoria'].unique():
    datos = df[df['categoria'] == categoria]['valor']
    print(f"\n{categoria}:")
    print(f"  Promedio: {datos.mean():.2f}")
    print(f"  Máximo: {datos.max()}")
    print(f"  Mínimo: {datos.min()}")

valor_maximo = df['valor'].max()
muestras_max = df[df['valor'] == valor_maximo]['nombre'].values

print(f"\n=== RESUMEN ===")
print(f"Valor máximo global: {valor_maximo}")
print(f"Encontrado en: {', '.join(muestras_max)}")
````

## Ejercicio 3.2: Clasificación Condicional

**Tarea:**
- Crea una nueva columna 'clasificacion' que clasifique los valores como:
  - "Muy Alto" si valor >= 160
  - "Alto" si 120 <= valor < 160
  - "Bajo" si valor < 120
- Muestra cuántos elementos hay en cada clasificación

````python
# Escribe tu código aquí
````

**Solución esperada:**

````python
# Solución
def clasificar(valor):
    if valor >= 160:
        return "Muy Alto"
    elif valor >= 120:
        return "Alto"
    else:
        return "Bajo"

df['clasificacion'] = df['valor'].apply(clasificar)

print("Cantidad por clasificación:")
print(df['clasificacion'].value_counts())
print("\nDatos clasificados:")
print(df[['nombre', 'valor', 'clasificacion']])
````

---

## Ejercicio 3.3: Visualización

**Tarea:**
- Crea un gráfico de barras que muestre el valor promedio por categoría
- Agrega título y etiquetas de ejes
- Muestra el gráfico

````python
# Escribe tu código aquí
````

**Solución esperada:**

````python
# Solución
promedio_categoria = df.groupby('categoria')['valor'].mean()
promedio_categoria.plot(kind='bar', title='Valor Promedio por Categoría', color='skyblue')
plt.xlabel('Categoría')
plt.ylabel('Valor Promedio')
plt.tight_layout()
plt.show()
````

---

## Ejercicio 3.4: Análisis de Variación

**Tarea:**
- Para cada categoría, calcula:
  - Cuántos datos hay
  - Promedio
  - Desviación estándar
  - Valor mínimo y máximo
- Crea una tabla resumen con esta información

````python
# Escribe tu código aquí
````

**Solución esperada:**

````python
# Solución
resumen = df.groupby('categoria')['valor'].agg([
    ('Cantidad', 'count'),
    ('Promedio', 'mean'),
    ('Desv. Est.', 'std'),
    ('Mínimo', 'min'),
    ('Máximo', 'max')
])
print("Resumen por Categoría:")
print(resumen)
````
