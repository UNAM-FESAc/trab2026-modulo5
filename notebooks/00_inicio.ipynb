## Parte 1: Tu Primer Código

La celda de abajo contiene código Python muy simple. Lee el código, luego haz clic en la celda y presiona **Shift + Enter** para ejecutarlo.

¿Qué crees que pasará?

## Parte 2: Trabajar con Variables

Una **variable** es como una caja donde guardamos información.

```python
# Crear variables
año = 2026
print("Estamos en el año", año)
```

### Explicación:
- `año = 2026` → creamos una variable llamada "año" con el valor 2026
- `print()` → mostramos el contenido

## Parte 3: Diferentes Tipos de Datos

Python puede trabajar con diferentes tipos de información:

```python
# Texto (string)
nombre = "María"
print(f"Mi nombre es {nombre}")

# Número entero (int)
edad = 25
print(f"Tengo {edad} años")

# Número decimal (float)
altura = 1.65
print(f"Mi altura es {altura} metros")

# Verdadero o Falso (bool)
es_estudiante = True
print(f"¿Soy estudiante? {es_estudiante}")
```

### Nota sobre `f"{variable}"`:
La letra `f` antes de las comillas permite incluir variables dentro del texto usando `{}`

## Parte 4: Operaciones Matemáticas

```python
# Suma
a = 10
b = 5
suma = a + b
print(f"{a} + {b} = {suma}")
```

```python
# Resta
resta = a - b
print(f"{a} - {b} = {resta}")
```

```python
# Multiplicación
multiplicacion = a * b
print(f"{a} × {b} = {multiplicacion}")
```

```python
# División
division = a / b
print(f"{a} ÷ {b} = {division}")
```

```python
# Potencia
potencia = a ** 2
print(f"{a} al cuadrado = {potencia}")
```

## Parte 5: Trabajar con Listas

Una **lista** es una colección de elementos.

```python
# Crear una lista
colores = ["rojo", "azul", "verde", "amarillo"]
print("Lista de colores:", colores)
```

```python
# Acceder a elementos (recuerda: contamos desde 0)
primer_color = colores[0]
print(f"El primer color es: {primer_color}")

segundo_color = colores[1]
print(f"El segundo color es: {segundo_color}")
```

```python
# Contar elementos
cantidad = len(colores)
print(f"Hay {cantidad} colores en la lista")
```

```python
# Agregar elemento
colores.append("naranja")
print("Después de agregar naranja:", colores)
```

## Parte 6: Bucles (Repetir Acciones)

A veces queremos repetir la misma acción varias veces. Para eso usamos bucles.

```python
# Bucle for: repetir para cada elemento
print("Mostrando cada color:")
for color in colores:
    print(f"  - {color}")
```

```python
# Bucle range: repetir un número de veces
print("Contando del 1 al 5:")
for numero in range(1, 6):
    print(numero)
```

## Parte 7: Condicionales (Si... entonces...)

A veces queremos que Python haga cosas diferentes según una condición.

```python
edad = 18

if edad >= 18:
    print("Eres un adulto")
else:
    print("Eres menor de edad")
```

```python
# Más opciones con elif (else if)
nota = 8

if nota >= 9:
    print("¡Excelente!")
elif nota >= 7:
    print("¡Muy bien!")
elif nota >= 5:
    print("Bien")
else:
    print("Necesitas mejorar")
```

## Parte 8: Crear Funciones

Una **función** es un conjunto de instrucciones que puedes reutilizar.

```python
# Definir una función simple
def saludar(nombre):
    print(f"¡Hola, {nombre}! Bienvenido.")

# Usar la función
saludar("Carlos")
saludar("Ana")
saludar("Pedro")
```

```python
# Función que devuelve un resultado
def calcular_promedio(notas):
    suma = sum(notas)
    cantidad = len(notas)
    promedio = suma / cantidad
    return promedio

# Usar la función
mis_notas = [8, 9, 7, 8]
mi_promedio = calcular_promedio(mis_notas)
print(f"Mi promedio es: {mi_promedio}")
```

## Resumen de Conceptos

| Concepto | Ejemplo | Uso |
|----------|---------|-----|
| Variable | `nombre = "Juan"` | Guardar datos |
| Lista | `[1, 2, 3]` | Guardar múltiples elementos |
| Bucle | `for i in range(5):` | Repetir acciones |
| Condicional | `if x > 0:` | Tomar decisiones |
| Función | `def mi_funcion():` | Reutilizar código |
| Print | `print(x)` | Mostrar resultados |
